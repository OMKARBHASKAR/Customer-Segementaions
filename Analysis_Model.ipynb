```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df['Education'].value_counts()  
```

```python
df['Marital_Status'].value_counts() 
```

```python
df['Dt_Customer']= pd.to_datetime(df['Dt_Customer'],dayfirst=True)
```

```python
df.info()
```

```python
df['Age'] = 2024 - df['Year_Birth']
```

```python
df['Age']
```

```python
df.head()
```

```python
df['Total_children'] = df['Kidhome'] + df['Teenhome']
```

```python
df['Total_children']
```

```python
spend_col = ["MntWines", "MntFruits", "MntMeatProducts", "MntFishProducts", "MntSweetProducts", "MntGoldProds"]
```

```python
df["Total_spend"] = df[spend_col].sum(axis=1)
```

```python
df["Total_spend"]
```

```python
df['customers_since'] = (pd.Timestamp("today") - df['Dt_Customer']).dt.days
```

```python
df['customers_since']
```

```python
sns.histplot(df['Age'], bins=20, kde=True)
plt.title('Age Distribution')
plt.show()
```

```python
sns.histplot(df['Income'], bins=30, kde=True)
plt.title('Income Distribution')
plt.show()
```

```python
sns.histplot(df['Total_spend'], bins=30, kde=True)
plt.title('Total Spend Distribution')
plt.show()
```

```python
sns.boxplot(x='Education', y='Income', data=df)
plt.xticks(rotation=45)
plt.title('Income by Education Level')
plt.show()
```

```python
sns.boxplot(x='Marital_Status', y='Total_spend', data=df)
plt.xticks(rotation=45)
plt.title('Total Spend by Marital Status')
plt.show()
```

```python
corr = df[[ 'Income','Age','Recency','Total_spend','NumWebPurchases','NumStorePurchases']].corr()
```

```python
sns.heatmap(corr, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```

```python
pivot_income = df.pivot_table(values='Income', index='Education', columns='Marital_Status', aggfunc='mean')
```

```python
pivot_income
```

```python
sns.heatmap(pivot_income, annot=True,fmt="0.f",cmap='YlGnBu')
plt.title('Average Income by Education and Marital Status') 
plt.show()
```

```python
sns.heatmap(pivot_income, annot=True, cmap='YlGnBu', linewidths=0.5)
plt.title('Average Income by Education and Marital Status') 
plt.show()
```
